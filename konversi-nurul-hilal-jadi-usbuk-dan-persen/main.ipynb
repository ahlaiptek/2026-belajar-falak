```python
import math

def hitung_usbu_hilal(derajat, menit, detik):
    """
    Menghitung usbu' (jari) dan persentase cahaya hilal
    berdasarkan data cahaya hilal dalam format DMS (Derajat, Menit, Detik).
    
    Ketentuan:
    - 1 jari (usbu') = 1°
    - Cahaya bulan maksimal = 12 jari (setara 100%)
    """
    
    print("=" * 50)
    print("📌 LANGKAH 1: Data Awal (Format DMS)")
    print("=" * 50)
    print(f"Derajat : {derajat}°")
    print(f"Menit   : {menit}'")
    print(f"Detik   : {detik}\"")
    print(f"Ditulis : {derajat}° {menit}' {detik}\"\n")
    
    print("=" * 50)
    print("📌 LANGKAH 2: Konversi Menit ke Desimal Derajat")
    print("=" * 50)
    menit_desimal = menit / 60
    print("Rumus   : menit ÷ 60")
    print(f"Hitung  : {menit} ÷ 60 = {round(menit_desimal, 5)}°\n")
    
    print("=" * 50)
    print("📌 LANGKAH 3: Konversi Detik ke Desimal Derajat")
    print("=" * 50)
    detik_desimal = detik / 3600
    print("Rumus   : detik ÷ 3600")
    print("Catatan : 1° = 60' dan 1' = 60\", maka 1° = 3600\"")
    print(f"Hitung  : {detik} ÷ 3600 = {round(detik_desimal, 5)}°\n")
    
    print("=" * 50)
    print("📌 LANGKAH 4: Menjumlahkan Semua Komponen")
    print("=" * 50)
    nilai_desimal = derajat + menit_desimal + detik_desimal
    print("Rumus   : derajat + menit_desimal + detik_desimal")
    print(f"Hitung  : {derajat} + {round(menit_desimal, 5)} + {round(detik_desimal, 5)}")
    print(f"Hasil   : {round(nilai_desimal, 5)}°\n")
    
    print("=" * 50)
    print("📌 LANGKAH 5: Menghitung Usbu' (Jari)")
    print("=" * 50)
    usbu_jari = math.ceil(nilai_desimal)
    print("Rumus   : nilai_desimal ÷ 1° (dibulatkan ke atas)")
    print("Catatan : 1 jari = 1° (0°60'), pembulatan ke atas")
    print(f"Hitung  : {round(nilai_desimal, 5)}° ÷ 1° = {round(nilai_desimal, 5)}")
    print(f"Dibulatkan ke atas menjadi = {usbu_jari} jari\n")
    
    print("=" * 50)
    print("📌 LANGKAH 6: Menghitung Persentase Cahaya")
    print("=" * 50)
    persen = (nilai_desimal / 12) * 100
    print("Rumus   : (nilai_desimal ÷ 12) × 100")
    print("Catatan : 12 jari = cahaya bulan maksimal (100%)")
    print(f"Hitung  : ({round(nilai_desimal, 5)} ÷ 12) × 100 = {round(persen, 5)}%")
    print(f"Dibulatkan menjadi = {round(persen)}%\n")
    
    print("=" * 50)
    print("✅ HASIL AKHIR")
    print("=" * 50)
    print(f"Cahaya Hilal   : {derajat}° {menit}' {detik}\"")
    print(f"Nilai Desimal  : {round(nilai_desimal, 5)}°")
    print(f"Usbu' (Jari)   : {usbu_jari} jari")
    print(f"Persentase     : {round(persen)}%")
    print("=" * 50)
    
    return {
        "nilai_desimal": round(nilai_desimal, 5),
        "usbu_jari": usbu_jari,
        "persen": round(persen)
    }


# Contoh penggunaan
hasil = hitung_usbu_hilal(5, 35, 50)
```
