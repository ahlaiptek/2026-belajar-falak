Kalau yang kamu maksud dalam ilmu falak, jaib (جيب) dan qaus (قوس) berkaitan dengan fungsi sinus dan arcsinus.
- Jaib (جيب) = sinus → dari qaus ke jaib.
- Qaus (قوس) = sudut/arc → dari jaib kembali menjadi sudut.
- Dalam Python: sin() untuk jaib, asin() untuk qaus.

Qaus -> Jaib


```python
import math

qaus = 30

jaib = math.sin(math.radians(qaus))

print("Qaus :", qaus, "°")
print("Jaib :", jaib)
```
