
   Qaus (قوس) itu apa?
Qaus secara bahasa berarti busur.
Dalam matematika/falak, qaus digunakan untuk menyatakan besar sudut.
Misalnya kita punya lingkaran:
             A
           / |
         /   |
       /     |
     /       |
   O---------B

Sudut:
AOB = 30 derajat
Maka 30° adalah qaus.

Jadi secara sederhana:
Qaus = ukuran sudut/busur yang kita masukkan.

Contoh:
- qaus = 0°
- qaus = 30°
- qaus = 45°
- qaus = 90°
- qaus = 120°


Dan dalam python kita menggunakan asin() untuk qaus

```python
import math

qaus = 30

jaib = math.sin(math.radians(qaus))

print("Qaus :", qaus, "°")
print("Jaib :", jaib)



# Intinya aku masih belum mengerti penulisan itu
```
