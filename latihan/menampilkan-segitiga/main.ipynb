```python
import matplotlib.pyplot as plt

# Koordinat titik
A = (0, 0)
B = (4, 0)
C = (4, 3)

# Hubungkan titik menjadi segitiga
x = [A[0], B[0], C[0], A[0]]
y = [A[1], B[1], C[1], A[1]]

plt.plot(x, y, marker="o")

# Label titik
plt.text(-0.2, -0.2, "A")
plt.text(4.1, -0.2, "B")
plt.text(4.1, 3, "C")

# Label sisi
plt.text(2, -0.4, "4")
plt.text(4.15, 1.5, "3")
plt.text(1.5, 1.7, "5")

plt.title("Segitiga Siku-siku")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.axis("equal")
plt.show()
```
